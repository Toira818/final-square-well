```python
#original

import numpy as np
import matplotlib.pyplot as plt

L = 2*np.pi #giving L a value
x_vals = np.linspace(0, L, 1000)  #evaluating psi at these x values
A = (2/L)**0.5 #normalisation constant

def psi(n,x):
    wf = A*np.sin(n*np.pi*x/L) #defining the wave function
    return wf

y_vals1 = psi(1, x_vals)
y_vals2 = psi(2, x_vals)
y_vals3 = psi(3, x_vals)
y_vals4 = psi(4, x_vals)


plt.plot(x_vals, y_vals1, 'r', linewidth = 3.0, label = "n = 1")
plt.plot(x_vals, y_vals2, 'b', label = "n = 2")
plt.plot(x_vals, y_vals3, 'k', label = "n = 3")
plt.plot(x_vals, y_vals4, 'g', label = "n = 4")


plt.legend()
plt.title("Wavefunction for Infinite Square Well")
plt.xlabel("x values")
plt.ylabel("psi(x)")
plt.grid()

```

```python
#probability density

import numpy as np
import matplotlib.pyplot as plt

L = 2*np.pi #giving L a value
x_vals = np.linspace(0, L, 1000)  #evaluating psi at these x values
A = (2/L)**0.5 #normalisation constant

def psi(n,x):
    wf = A*np.sin(n*np.pi*x/L) #defining the wave function
    return wf


for n in range(1,4):
    y_vals = psi(n, x_vals)**2
    plt.plot(x_vals, y_vals, label = 'n = {}'.format(n))
    
plt.title("Probability Density for Infinite Square Well")
plt.xlabel("x values")
plt.ylabel("psi(n,x)^2")
plt.legend()





```

```python
#energy

import numpy as np
import matplotlib.pyplot as plt

#constants AMU
L = 2*np.pi  
hbar = 1 
m = 1  


def energy(n):
    E = hbar**2*np.pi**2*n**2/(2*m*L**2)
    return E

#draws a horizontal line across the well width (0 to L)
x_vals = np.linspace(0, L, 1000)

for n in range(1, 4):
    E_n = energy(n)
    # Creates a constant array of the energy value to plot a flat line
    y_vals = np.full_like(x_vals, E_n) 
    plt.plot(x_vals, y_vals, label='n = {} (E = {:.3f})'.format(n, E_n))

plt.title("Discrete Energy Levels of Infinite Square Well")
plt.xlabel("x_values")
plt.ylabel("Energy")
plt.legend()
plt.grid()



```
